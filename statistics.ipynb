**for find hypothesis test**

```python
raw_data=dataset['Age']
p_mean=raw_data.mean()
p_std=raw_data.std()
sample_size=35
sample_mean=raw_data.sample(n=35,random_state=4).mean()
```

```python
z_value=(sample_mean-p_mean)/(p_std/np.sqrt(sample_size))
p_value=2*(1-norm.cdf(z_value))
if p_value>0.05:
  print("accept null hypothesis")
else:
  print("reject null hypothesis")
```
